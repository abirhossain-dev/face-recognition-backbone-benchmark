```python
from sklearn.datasets import fetch_lfw_people

# Load dataset (resize and min_faces_per_person are optional)
lfw = fetch_lfw_people(min_faces_per_person=20, resize=0.5, color=False)

print("Images shape:", lfw.images.shape)   # (n_samples, height, width)
print("Data shape:", lfw.data.shape)       # (n_samples, height*width)
print("Targets shape:", lfw.target.shape)  # (n_samples,)
print("Target names:", lfw.target_names[:10])
```

```python
import os

# Where to store the dataset locally
os.makedirs("lfw_data", exist_ok=True)

!kaggle datasets download -d jessicali9530/lfw-dataset -p lfw_data

# Unzip
import shutil
zip_files = [f for f in os.listdir("lfw_data") if f.endswith(".zip") or f.endswith(".7z") or f.endswith(".tar") or f.endswith(".tgz")]
print("Found archives:", zip_files)

for z in zip_files:
    shutil.unpack_archive(os.path.join("lfw_data", z), "lfw_data")

print("Files in lfw_data:", os.listdir("lfw_data"))
```
